**EXERCISE 1:** Write a Python program that checks whether a number is positive or negative.

```python
num = float(input("Enter a number: "))
if num > 0:
   print("Number is a positive number.")
elif num < 0:
   print("Number is a negative number.")
else:
   print("Number is zero.")
```

**EXERCISE 4:** Write a function that takes two numbers as input and returns their sum.

```python
num1 = float(input("Enter 1st number: "))
num2 = float(input("Enter 2nd number: "))
def sumOfTwoNumbers(num1, num2):
    totalSum = num1 + num2
    return totalSum
print("Sum of", num1, "and", num2, ":", sumOfTwoNumbers(num1, num2))
```

**EXERCISE 5:** Create a class named &#39;Car&#39; with attributes brand and year, and a method to display them.

```python
class Car:
  def __init__(self, brand, year, name, model):
    self.brand = brand
    self.year = year
    self.name = name
    self.model = model
  def display(self):
    print("Name:", self.name, ", Brand:", self.brand, ", Model:", self.model, ", Year:", self.year)

car1 = Car('Audi', 2019, 'A3 Audi', 'Q5')
car2 = Car('Toyata', 2020, 'Civics', '74')
print("Car 1")
car1.display()
print("Car 2")
car2.display()
```

**EXERCISE 6:** Matrices

```python
import numpy as np

print('Task 1 ')
matrix1 = np.array([[2, 3], [4, 5]])
Task1 = 2 * matrix1
print(Task1)

print('Task 2')
matrix2 = np.array([2, 3])
matrix3 = np.array([[4],[5]])
Task2 = np.dot(matrix2, matrix3)
print(Task2)

print('Task 3')
matrix4 = np.array([[1, 2], [3, 4]])
Task3 = np.dot(matrix2, matrix4)
print(Task3)

print('Task 4')
Task4 = matrix4.T
print(Task4)

print("Task 5")#get [1,2] and [3]
Task5Slice1 = matrix4[0, :]
Task5Slice2 = matrix4[1,0:1]
print(Task5Slice1)
print(Task5Slice2)

print("Task 6")
Task6 = np.linalg.inv(matrix4)
print(Task6)
```
